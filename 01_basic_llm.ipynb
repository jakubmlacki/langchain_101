```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, AIMessage

from dotenv import load_dotenv

# Load environmental variables from .env
load_dotenv()

```

```python
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.7)
res = None

def ask_city_activity(activity: str, city: str) -> str:
    global res

    prompt = (
        f"I like {activity}. Check if in the city of {city} there are available activities connected with those activities? "
        "Provide 3 examples"
    )
    res = llm.invoke(prompt)  # res is complex object with multiple interesting attributes
    return res
```

```python

result = ask_city_activity("climbing", "San Francisco")
```

```python

print(result.content)
```

```python
# Another call... is not aware of previous messages
print(llm.invoke("Are there any other propositions you can suggest?").content)
```

Add message memory!

```python
messages = []

def ask_city_activity_with_memory(input_prompt: str) -> str:
    global messages
    messages.append(HumanMessage(input_prompt))
    ai = llm.invoke(messages)
    messages.append(AIMessage(ai.content))
    return ai.content
```

```python
prompt = (
        "I like climbing. Check if in the city of San Francisco there are available activities connected with those activities?"
        "Provide 3 propositions"
    )
result = ask_city_activity_with_memory(prompt)
```

```python
print(result)
```

```python
prompt = (
        f"Are there any other propositions you can suggest?"
    )
print(ask_city_activity_with_memory(prompt))
```
