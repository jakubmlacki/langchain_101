```python
from pydantic import BaseModel, Field
from typing import List

from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

load_dotenv()
```

```python
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.7)
```

```python
class CityActivities(BaseModel):
    city: str
    interests: List[str]
    suggested_activities: List[str] = Field(description="Suggested, short activities (3-5).")
    confidence: float = Field(ge=0, le=1)

structured_llm = llm.with_structured_output(CityActivities)

resp = structured_llm.invoke(
    "I like food and museums. I will visit Poznań — show me propositions."
)
resp
```
